```python
import pandas as pd
import yaml
import glob

folders_path = r"../Data/data"
folder_pattern = f"{folders_path}\*"
all_folders = glob.glob(folder_pattern)
stock_combined_df = pd.DataFrame()

for folder in all_folders:

    file_pattern = f"{folder}\*.yaml"
    all_files = glob.glob(file_pattern)
    
    for curr_file in all_files:
        with open(curr_file, 'r') as file:
            yaml_data = yaml.safe_load(file)
        df = pd.DataFrame(yaml_data)
        stock_combined_df = pd.concat([stock_combined_df, df])

stock_combined_df

```

```python
unique_stocks = stock_combined_df['Ticker'].unique()
path = r"../Data/stocks_csv"
for stock in unique_stocks:
    unique_stock_df = stock_combined_df[stock_combined_df['Ticker'] == stock]
    unique_stock_df.to_csv(f"{path}\{stock}.csv", index=False)


```

```python
from datetime import datetime, date

data = []
for index in stock_combined_df.index:
    row = stock_combined_df.loc[index].values
    row[0] = str(row[0])
    row[1] = float(row[1])
    row[2] = datetime.strptime(row[2], '%Y-%m-%d %H:%M:%S')
    row[3] = float(row[3])
    row[4] = float(row[4])
    year_month = row[5].split('-')
    formattted_month = date(int(year_month[0]),int(year_month[1]),1)
    row[5] = formattted_month
    row[6] = float(row[6])
    row[7] = int(row[7])
    data.append(tuple(row))

query = "insert into stock_exchange_details values(%s,%s,%s,%s,%s,%s,%s,%s)"
cursor.executemany(query,data)
connection.commit()
```
